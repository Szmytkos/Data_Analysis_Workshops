# Report: Univariate Analysis

##### In this quick univariate analysis, we will explore the distribution and data quality of the rating variable.

```python
by_room = airbnb.groupby('room_type')[['rating']].agg(['mean', 'median', 'count', 'min', 'max']).round(3)
by_room.columns = ['_'.join(col) for col in by_room.columns]
by_room = by_room.reset_index()
by_room
```

First, let's visualize the differences in median ratings using a bar plot.

```python
plt.figure(figsize=(8, 5))
ax = sns.barplot(x=by_room['room_type'], y=by_room['rating_median'], hue=by_room['room_type'], palette='viridis')
plt.ylim(3.95, 4.05)
for container in ax.containers:
    ax.bar_label(container, fmt='%.3f', padding = 3)
plt.xlabel('Room Type')
plt.ylabel('Rating')
plt.show()
```

As we see difference between mean in rating by room types is very small

Let's use a violin plot to examine the distribution and density of the ratings. This will show us if the actual spread of scores differs depending on the room type.

```python
plt.figure(figsize=(10,6))
sns.violinplot(x='room_type', y='rating', hue='room_type', data=airbnb, palette='viridis')
plt.ylabel('Rating')
plt.xlabel('Room Type')
plt.show()
```

As we can see from the violin plots, the overall distribution of ratings is very similar for all room types. The data is highly concentrated around the scores of 4.0, 4.5, and 5.0

# Report: Bivariate Analysis


We will use a Spearman correlation heatmap to identify any relationships between our numerical features. Spearman is chosen here because it handles outliers.

```python
num_airbnb = airbnb.select_dtypes(['int64', 'float64'])
corr_spear = num_airbnb.corr(method='spearman')
plt.figure(figsize=(10,8))
sns.heatmap(corr_spear, annot=True, cmap='coolwarm')
plt.show()
```

The heatmap reveals a few interesting insights. The first one is the relationship between rating and price. The correlation value is near zero (0.00024), which means that the price of a listing has absolutely no influence on the final rating given by guests.

Another notable finding is the moderate negative correlation -0.43 between price and longitude. This indicates a clear geographic trend as we move further in a specific direction, the rental rices drop significantly.
